```python

# ── Nested loop: multiplication table ────────────────────────────────
print('Multiplication Table (1–5)\n')
print('   ', end='')              # header spacing
for col in range(1, 6):
    print(f'{col:4}', end='')     # column headers
print()                           # newline after header
print('─' * 25)

for row in range(1, 6):           # outer loop: rows 1 to 5
    print(f'{row:2} |', end='')   # row label
    for col in range(1, 6):       # inner loop: columns 1 to 5
        print(f'{row*col:4}', end='')  # compute and print product
    print()
```

```python

# ── Functions for geometric formulas ────────────────────────────────
import math   # import Python's math module for π and √

def circle_area(r):
    """Returns area of a circle given radius r: A = πr²"""
    return math.pi * r ** 2       # math.pi gives precise value of π

def circle_circumference(r):
    """Returns circumference: C = 2πr"""
    return 2 * math.pi * r

def triangle_area(base, height):
    """Returns area of a triangle: A = ½ × base × height"""
    return 0.5 * base * height

# Test the functions
r = 5
print(f'Circle  Area (r={r})        : {circle_area(r):.4f}')
print(f'Circle  Circumference (r={r}): {circle_circumference(r):.4f}')
print(f'Triangle Area (b=6, h=4)    : {triangle_area(6, 4):.4f}')
```

```python
double  = lambda x: x * 2          # f(x) = 2x
cube    = lambda x: x ** 3         # f(x) = x³
celsius_to_fahrenheit = lambda c: (c * 9/5) + 32  # temperature conversion

print(double(7))                    # 14
print(cube(4))                      # 64
print(celsius_to_fahrenheit(100))   # 212.0 — boiling point
```

```python

# ── Factorial: n! = 1 × 2 × 3 × ... × n ─────────────────────────────
n = 6           # compute 6!
factorial = 1   # start with product = 1 (not 0, as we are multiplying)

for i in range(1, n + 1):   # i goes 1, 2, 3, 4, 5, 6
    factorial *= i           # multiply accumulator by i
    print(f'  Step {i}: {i}! so far = {factorial}')

print(f'{n}! = {factorial}')
```

```python
# Fibonacci: each number = sum of the two before it → 0, 1, 1, 2, 3, 5, 8, ...
a, b = 0, 1       # initialise first two Fibonacci numbers
num_terms = 100   # generate the first 100 terms

print(f'First {num_terms} Fibonacci numbers:')
for _ in range(num_terms): # Loop for the specified number of terms
    print(a, end='  ')     # print current value (end='  ' keeps on same line)
    a, b = b, a + b        # calculate the next Fibonacci number
```
