```python
import os
from datetime import datetime
import random
def get_readings_by_location(readings_list, target_location):
    return [reading for reading in readings_list
            if reading["LOCATION"] == target_location]

LOCATIONS = ["Hyderabad", "Delhi", "Mumbai", "Chennai", "Bengaluru"]
DEVICE_TYPES = ["Temperature", "Humidity", "Pressure", "Motion"]

target = input("Enter target location: ")

ar = []

for _ in range(100):
    now = datetime.now()

    reading = {
        "TIME": now.strftime("%y-%m-%d %H:%M:%S"),
        "DEVICE_ID": "DEV" + str(random.randint(1, 100)),
        "VALUE": round(random.random(), 2),
        "LOCATION": random.choice(LOCATIONS),
        "DEVICE_TYPE": random.choice(DEVICE_TYPES)
    }

    ar.append(reading)
print()

result = get_readings_by_location(ar, target)

print(result)
```
