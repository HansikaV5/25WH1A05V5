```python
import numpy as np
import random as random
```

```python
# Task 1: Array Creation Shortcuts
print("\n--- Task 1: Array Creation Shortcuts ---")
arr1_1 = np.arange(0, 10, 2)
arr1_2 = np.ones(9).reshape(3,3)
arr1_3 = np.array([7,7,7,7,7,7,7,7]).reshape(2,4)
arr1_4 = np.array([random.randint(10, 50) for _ in range(25)]).reshape(5, 5)

print("arr1_1 (first 10 elements):", arr1_1[:10])
print("arr1_2:\n", arr1_2)
print("arr1_3:\n", arr1_3)
print("arr1_4 (first 2 rows):\n", arr1_4[:2])
```
